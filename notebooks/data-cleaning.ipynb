```python
import pandas as pd
```

```python
bios = pd.read_csv("../data/bios.csv")
```

```python
bios.head()
```

## Getting rid of bullet points in used names

```python
df = bios.copy()
```

```python
df["Name"] = df["Used name"].str.replace("•"," ")
```

```python
df.head()
```

## Spliting the measurement column in height & weight column

```python
df[["Height_cm","Weight_kg"]] = df["Measurements"].str.split("/",expand=True)

# Removing cm and converting to numeric value
df["Height_cm"] = df["Height_cm"].str.strip(' cm')
df["Height_cm"] = pd.to_numeric(df["Height_cm"],errors="coerce")
```

```python
df.head(3)
```

```python
# Removing kg and converting into numeric value
df["Weight_kg"] = df["Weight_kg"].str.strip(' kg')
df["Weight_kg"] = pd.to_numeric(df["Weight_kg"],errors="coerce")
```

```python
df.head(3)
```

## Parsing Birth and Death Information


```python
date_pattern = r'(\d+ \w+ \d{4}|\d{4})'
df['born_date'] = df["Born"].str.extract(date_pattern)
```

```python
df.head(3)
```

```python
df['Died_date'] = df["Died"].str.extract(date_pattern)
```

```python
df.head(3)
```

```python
# Converting into date format
df["born_date"] = pd.to_datetime(df["born_date"], errors="coerce")
df["Died_date"] = pd.to_datetime(df["Died_date"], errors="coerce")
```

```python
df.head(3)
```

## Extracting city, region, and country from the Born column

```python
location_pattern = r'in ([\w\s-]+), ([\w\s-]+) \((\w+)\)'
df["Born"].str.extract(location_pattern)
```

```python
df[['born_city','born_region','born_country']] = df ['Born'].str.extract(location_pattern,expand=True)
```

```python
df.head(3)
```

```python
import pandas as pd
pd.set_option('display.max_colwidth',None)
df.value_counts("born_city",ascending=False).head(10)
```

## Removing unnecessary columns

```python
df.columns
```
